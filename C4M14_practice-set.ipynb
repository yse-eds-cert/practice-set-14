### Extension: Optical Character Recognition (OCR)

So far in this week's lectures and this practice set, you have been working with a **binary** classification model. Now, we will briefly go through an example of a **multiclass** classification model, that also happens to be a useful skill to have if you want to work with text data - Optical Character Recognition, or OCR.

(As this was not covered in the lecture material, we will not grade you for credit for this section.)

OCR refers to the use of machine learning classifiers to convert image data into text. In the same way we learnt that text data can be converted into a vector of tokens, with each token represented by 1/0 for whether the token is present in each document, image data can be converted into a matrix of pixels, with each pixel represented by a numeric value. In the simplest case, a black and white photo can be represented by a matrix of 1s and 0s, with 1s representing black pixels and 0s representing white pixels. Then, in the same way that the binary classifier we just built matches the numeric token vector to a topic label for that text, an OCR multiclass classifier can match the numeric pixel matrix to a character label for that image. You can read more about OCR [here](https://towardsdatascience.com/optical-character-recognition-with-knn-classifier-10fd220ed797/), [here](https://www.affinda.com/blog/machine-learning-ocr) and [here](https://aws.amazon.com/what-is/ocr).

Luckily, both [R](https://cran.r-project.org/web/packages/tesseract/vignettes/intro.html) and [Python](https://www.datacamp.com/tutorial/optical-character-recognition-ocr-in-python-with-pytesseract) have plug-and-play packages for OCR that do all the classification work for you, so (most of the time) you won't have to worry about getting too into the weeds with the machine-learning theory. We will work through a short exercise using one such package.

You need to access the text from those PDFs somehow! Thankfully, R's `tesseract` and `pdftools` packages can perform this task for you.
